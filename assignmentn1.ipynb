```python
message = "&&&**$gnirtS PLIO!!@1234"
#core
core=message[6:-7]
#split
words = core.split(" ")
#reverse the first word
first_word = words[0][::-1]
#Replace shifted vowels
second_word = words[1]
second_word = second_word.replace("I", "E")
second_word = second_word.replace("O", "U")
#decoded message
decoded = first_word + " " + second_word
print(decoded)
```

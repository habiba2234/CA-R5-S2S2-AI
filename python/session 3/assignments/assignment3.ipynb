```python
#problem1
def multiplication_number(number):
    """
    Brief description:
    This function prints the multiplication table
    of a given number from 1 to 10.
    Args:
        number (int): The number used to generate
        the multiplication table.
    Returns:
        None: Prints the multiplication table.
    Examples:
        >>> multiplication_number(9)
        9 * 1 = 9
        9 * 2 = 18
        9 * 3 = 27
        ...
        9 * 10 = 90
    """

    for i in range(1, 11):
        print(number, "*", i, "=", number * i)
```

```python
multiplication_number(9)
```

```python
#problem2
def is_prime(number):
    """
    Brief description:
    This function checks whether a given number is prime.
    Args:
        number (int): The number to check.
    Returns:
        bool: True if the number is prime,
        otherwise False.
    Examples:
        >>> is_prime(7)
        True

        >>> is_prime(9)
        False
    """
    if number < 2:
        return False
    for i in range(2, number):
        if number % i == 0:
            return False
    return True
```

```python
def twin_primes():
    """
    Brief description:
    This function prints all twin prime numbers
    less than 1000.
    Args:
        None
    Returns:
        None: Prints the twin prime numbers.

    Examples:
        >>> twin_primes()
        3 and 5
        5 and 7
        11 and 13
        ...
    """
    for i in range(3, 1000, 2):
        if is_prime(i) and is_prime(i + 2):
            print(i, "and", i + 2)
```

```python
twin_primes()
```

```python
#problem3
def prime_factor(number):
    """
    Brief description:
    This function finds the prime factors of a given number.
    Args:
        number (int): The number whose prime factors
        will be found.
    Returns:
        list: A list containing the prime factors
        of the given number.
    Examples:
        >>> prime_factor(56)
        [2, 2, 2, 7]
    """
    factors = []
    divisor = 2
    while number > 1:
        if number % divisor == 0:
            factors.append(divisor)
            number = number // divisor
        else:
            divisor += 1
    return factors
```

```python
print(prime_factor(12))
```

```python
#problem4
def decToBin(number):
    """
    Brief description:
    This function converts a decimal number
    into its binary representation.
    Args:
        number (int): The decimal number to convert.
    Returns:
        str: The binary representation of the number.
    Examples:
        >>> decToBin(11)
        '1011'
    """
    binary = ""
    while number > 0:
        remainder = number % 2
        binary = str(remainder) + binary
        number = number // 2
    return binary
```

```python
print(decToBin(11))
```

```python
#problem5
def perfectNums(start, end):
    """
    Brief description:
    This function prints all perfect numbers
    within a given range.
    Args:
        start (int): The starting value of the range.
        end (int): The ending value of the range.

    Returns:
        None: Prints the perfect numbers in the range.

    Examples:
        >>> perfectNums(1, 100)
        6
        28
    """
    for number in range(start, end):
        divisor_sum = 0

        for i in range(1, number):
            if number % i == 0:
                divisor_sum += i

        if divisor_sum == number:
            print(number)
```

```python
perfectNums(1, 100)
```

```python
#oop--->
# problem1
class MultiplicationTable:
    """
    This class generates the multiplication table
    of a given number.
    """
    def __init__(self, number):
        self.number = number
    def multiplication(self):
        """
        Prints the multiplication table from 1 to 10.
        """
        for i in range(1, 11):
            print(self.number, "*", i, "=", self.number * i)
table = MultiplicationTable(9)
table.multiplication()
```

```python
# problem2
class TwinPrimes:
    """
    This class finds twin prime numbers less than 1000.
    """
    def __init__(self, limit):
        self.limit = limit
    def is_prime(self, number):
        """
        Checks whether a number is prime.
        """
        if number < 2:
            return False
        for i in range(2, number):
            if number % i == 0:
                return False
        return True
    def find_twin_primes(self):
        """
        Prints all twin prime numbers less than the given limit.
        """
        for i in range(3, self.limit, 2):
            if self.is_prime(i) and self.is_prime(i + 2):
                print(i, "and", i + 2)
twin = TwinPrimes(1000)
twin.find_twin_primes()
```

```python
# problem3
class PrimeFactors:
    """
    This class finds the prime factors of a given number
    """
    def __init__(self, number):
        self.number = number
    def find_factors(self):
        """
        Finds and returns the prime factors of the number
        """
        factors = []
        divisor = 2
        while self.number > 1:
            if self.number % divisor == 0:
                factors.append(divisor)
                self.number = self.number // divisor
            else:
                divisor += 1
        return factors
factor = PrimeFactors(12)
print(factor.find_factors())
```

```python
# problem4
class DecimalToBinary:
    """
    This class converts a decimal number to binary
    """
    def __init__(self, number):
        self.number = number
    def convert(self):
        """
        Converts decimal number to a binary number
        """
        binary = ""

        while self.number > 0:
            remainder = self.number % 2
            binary = str(remainder) + binary
            self.number = self.number // 2
        return binary
number = DecimalToBinary(15)
print(number.convert())
```

```python
# problem5
class PerfectNumbers:
    """
    This class finds perfect numbers within a given range
    """
    def __init__(self, start, end):
        self.start = start
        self.end = end
    def find_perfect_numbers(self):
        """
        Prints all perfect numbers in the given range
        """
        for number in range(self.start, self.end):
            divisor_sum = 0
            for i in range(1, number):
                if number % i == 0:
                    divisor_sum += i

            if divisor_sum == number:
                print(number)
perfect = PerfectNumbers(1, 100)
perfect.find_perfect_numbers()
```
